# MongoDB CRUD Operations – Simple Python Notebook

This notebook uses **PyMongo** to perform the four basic CRUD operations.

- **Create** – add a trainee
- **Read** – show all trainees
- **Update** – change a trainee's age
- **Delete** – remove a trainee

**Before starting:** Make sure MongoDB Community Server is installed and running on your computer.

## 2. Connect to MongoDB
`MongoClient` connects Python to the MongoDB server running on this computer.

```python
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017/")
print("Connected to MongoDB!")
```

## 3. Select database and collection

- Database name: `training_center`
- Collection name: `trainees`

A MongoDB collection is similar to a table. A document is similar to a Python dictionary.

```python
db = client["training_center"]
trainees = db["trainees"]

print("Database and collection selected!")
```

## 4. CREATE – Add one trainee

`insert_one()` adds one dictionary (document) to the collection.

Change the name, age, or course and run the cell to add another trainee.

```python
trainee = {"name": "Veera", "age": 30, "course": "Python"}

trainees.insert_one(trainee)
print("Trainee added!")
```

## 5. READ – Show all trainees

`find()` gets the documents from the collection.

The loop displays each trainee. We use `trainee` as the variable for one document at a time.

```python
for trainee_doc in trainees.find():
    print(trainee_doc["name"], trainee_doc["age"], trainee_doc["course"])
```

## 6. UPDATE – Change a trainee's age

`update_one()` finds a trainee using the name and changes the age.

- First dictionary: which trainee to find.
- Second dictionary: what to change.
- `$set`: update the given field.

```python
result = trainees.update_one(
    {"name": "Veera"},
    {"$set": {"age": 31}}
)

print("Matched:", result.matched_count)
print("Modified:", result.modified_count)

if result.matched_count == 0:
    print("No trainee found with the name Veera.")
else:
    print("Trainee update operation completed!")
```

### Check the updated trainee

```python
updated_trainee = trainees.find_one({"name": "Veera"})

if updated_trainee:
    print(updated_trainee)
else:
    print("No trainee found with the name Veera.")
```

## 7. DELETE – Remove a trainee

`delete_one()` removes the first document that matches the name.

**Be careful:** This permanently deletes the matching document.

```python
trainees.delete_one({"name": "Veera"})
print("Trainee deleted!")
```

### Check all remaining trainees

```python
for trainee_doc in trainees.find():
    print(trainee_doc["name"], trainee_doc["age"], trainee_doc["course"])
```

## 8. Quick revision

| Operation | PyMongo method | Meaning |
|---|---|---|
| Create | `insert_one()` | Add a trainee |
| Read | `find()` | Get all trainees |
| Update | `update_one()` | Change trainee details |
| Delete | `delete_one()` | Remove a trainee |

**Remember:** If you run the CREATE cell repeatedly, MongoDB adds another document each time. If you see a connection error, check that MongoDB Server is running.
