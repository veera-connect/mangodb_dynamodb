```python
import json

# Fetch data from MongoDB
data = list(trainees.find({}, {"_id": 0}))

# Send data to Groq
response = llm.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[
        {
            "role": "system",
            "content": "Answer questions using only the given MongoDB data."
        },
        {
            "role": "user",
            "content": f"""
            MongoDB Data:
            {json.dumps(data, default=str)}

            Question: Show all trainees and explain their details.
            """
        }
    ]
)

print(response.choices[0].message.content)
```
